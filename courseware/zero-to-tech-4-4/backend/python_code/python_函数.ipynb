一、 函数的定义与调用
Python 使用 def 关键字来定义函数，函数体必须缩进。


```python
def 函数名(参数1, 参数2):
    """这是文档字符串，用于说明函数的功能（可选）"""
    函数体（要执行的代码逻辑）
    return 返回值（可选，如果没有return，默认返回None）
```

```python
# 1. 定义函数
def get_sum(a, b):
    """计算两个数的和"""
    result = a + b
    return result

# 2. 调用函数
total = get_sum(3, 5)
print(total)  # 输出：8
```

Python 的参数传递极其灵活，支持以下 4 种方式：
1. 位置参数（最基础）
按顺序一一对应传递，不能错位。

```python
def print_info(name, age):
    print(f"姓名：{name}，年龄：{age}")

print_info("Alice", 25)  # 正确
```

2. 关键字参数
调用时通过 参数名=值 传入，顺序可以打乱，适合参数多的场景。

```python
print_info(age=25, name="Alice")  # 正确，效果同上
```

3. 默认参数
定义时给参数一个默认值，调用时如果不传，就使用默认值。
⚠️ 注意：默认参数必须放在非默认参数的后面。

```python
def greet(name, msg="早上好"):
    print(f"{name}，{msg}!")

greet("小明")          # 输出：小明，早上好!
greet("小红", "下午好") # 覆盖默认值，输出：小红，下午好!
```

4. 可变参数（不定长参数）
当你不知道调用者会传多少个参数时使用：
*args：接收任意数量的位置参数，打包成元组 (tuple)。
**kwargs：接收任意数量的关键字参数，打包成字典 (dict)。

```python
def show_info(*args, **kwargs):
    print("位置参数:", args)
    print("关键字参数:", kwargs)

show_info(1, 2, name="Tom", age=20)
# 输出：
# 位置参数: (1, 2)
# 关键字参数: {'name': 'Tom', 'age': 20}
```

四、 函数的返回值 (return)
如果没有 return 语句，函数执行完毕后默认返回 None。
Python 支持一次返回多个值，底层是通过打包成“元组 (tuple)”实现的。

```python
def get_coordinates():
    x = 10
    y = 20
    return x, y  # 实际返回的是元组 (10, 20)

# 使用解包接收
x, y = get_coordinates()
print(f"x = {x}, y = {y}")  # 输出：x = 10, y = 20
```

五、 变量作用域（LEGB 规则）
在 Java 中，类成员变量和局部变量区分严格。Python 的作用域遵循 LEGB 规则：

L (Local)：函数内部局部变量。

E (Enclosing)：嵌套函数的外层函数局部变量。

G (Global)：模块级别的全局变量。

B (Built-in)：Python 内置的函数（如 print, len）。

⚠️ 避坑指南：在函数内部，你可以直接读取全局变量，但如果想要修改全局变量，必须使用 global 关键字显式声明，否则 Python 会把它当成一个新的局部变量。

题目1：学生成绩处理器
定义一个函数 process_scores(name, *scores)。
name 是位置参数，*scores 接收任意数量的成绩。
函数内部计算这些成绩的平均分。
返回一个包含 name 和 平均分 的元组。
调用该函数，并使用解包接收返回值，打印结果。

```python
def process_scores(name, *scores):
    """
    处理学生成绩：计算平均分并返回结果
    :param name: 学生姓名 (位置参数)
    :param scores: 任意数量的成绩 (可变参数，打包为元组) 
    :return: (姓名, 平均分) 的元组
    """
    # 1. 防御性编程：防止没有传入成绩导致除零错误
    if not scores:
        return name, 0.0
    
    # 2. 计算平均分：sum() 求和，len() 获取长度
    avg = sum(scores) / len(scores)
    
    # 3. 返回多个值：Python 会自动将其打包为元组
    return name, round(avg, 2)  # round() 保留2位小数

# --- 调用与解包 ---
student_name, student_avg = process_scores("Alice", 90, 85, 92, 78)
print(f"学生: {student_name}, 平均分: {student_avg}")

# 也可以只传姓名，不传成绩（测试防御性逻辑）
name2, avg2 = process_scores("Bob")
print(f"学生: {name2}, 平均分: {avg2}")
```

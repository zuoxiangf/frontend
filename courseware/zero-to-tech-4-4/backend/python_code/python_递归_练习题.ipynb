题目一：计算阶乘
题目描述：编写一个函数，计算正整数 n 的阶乘（ n!）。阶乘的定义为： n!=n×(n−1)×(n−2)×⋯×1 且 0!=1 。

```python
# 1. 先定义函数（先定义，后调用）
def factorial(n):
    """
    计算n的阶乘
    """
    # 基线条件：0和1的阶乘都是1（防止输入0导致无限递归）
    if n == 0 or n == 1:
        return 1
    else:
        # 递归推进：n * (n-1)的阶乘，必须有 return 才能把结果传出来
        return n * factorial(n - 1)

# 2. 后执行主逻辑
n = int(input("请输入一个整数："))
result = factorial(n)

print(f"当前输入的值为：{n}，阶乘为：{result}")
```

题目二：斐波那契数列
题目描述：编写一个函数，返回斐波那契数列的第 n 项。斐波那契数列的定义为： F(0)=0,F(1)=1 ，当 n>1 时， F(n)=F(n−1)+F(n−2) 。

```python
def func(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return func(n-1) + func(n-2)

n = int(input("请输入一个整数："))
result = func(n)

print(f"当前输入的值为：{n}，斐波那契数为：{result}")
```

题目三：汉诺塔问题
题目描述：汉诺塔是一个经典的递归问题。假设有 A、B、C 三根柱子，A 柱子上有 n 个大小不同的圆盘，按照从大到小的顺序排列。目标是将所有圆盘移动到 C 柱子上，且在移动过程中，任何时候大盘都不能放在小盘之上。编写一个函数打印出移动的步骤。

```python
def hanoi(n,source,auxiliary,target):
    if n == 1:
        print(f"将 圆盘1 从 {source} -> {target}")
        return
    hanoi(n-1,source,target,auxiliary)

    print(f"将 圆盘{n} 从 {source} -> {target}")

    hanoi(n-1,auxiliary,source,target)
    
hanoi(3, 'A', 'B', 'C') 
```

```python
def hanoi(n, source, auxiliary, target):
    """
    解决汉诺塔问题
    n: 圆盘数量
    source: 源柱子
    auxiliary: 辅助柱子
    target: 目标柱子
    """
    # 基线条件：只有一个圆盘时，直接从源移动到目标
    if n == 1:
        print(f"将圆盘 1 从 {source} 移动到 {target}")
        return
    # 递归步骤：
    # 1. 将上面 n-1 个圆盘从 source 移动到 auxiliary (借助 target)
    hanoi(n - 1, source, target, auxiliary)
    # 2. 将第 n 个圆盘从 source 移动到 target
    print(f"将圆盘 {n} 从 {source} 移动到 {target}")
    # 3. 将 n-1 个圆盘从 auxiliary 移动到 target (借助 source)
    hanoi(n - 1, auxiliary, source, target)

# 测试 (移动3个圆盘)
hanoi(3, 'A', 'B', 'C')
```

题目五：计算列表元素之和
题目描述：编写一个函数，计算列表中所有数字的总和，使用递归而不是循环。

```python
def sum(list):
    if not list:
        return 0
    else:
        return list[0] + sum(list[1:])

 # 测试
print(sum([1, 2, 3, 4, 5]))  # 输出: 15       
```

```python
def sum_list(lst):
    """
    递归计算列表中所有元素的和
    """
    # 基线条件：如果列表为空，返回0
    if not lst:
        return 0
    # 递归条件：列表的第一个元素 + 剩余列表的和
    else:
        return lst[0] + sum_list(lst[1:])

# 测试
print(sum_list([1, 2, 3, 4, 5]))  # 输出: 15
```

题目六：反转字符串
题目描述：编写一个函数，递归地反转给定的字符串。

```python
def reverse_string(s):
    """
    递归反转字符串
    """
    # 基线条件：空字符串或单字符字符串，直接返回
    if len(s) <= 1:
        return s
    # 递归条件：反转剩余部分 + 第一个字符
    else:
        return reverse_string(s[1:]) + s[0]

# 测试
print(reverse_string("hello"))  # 输出: olleh
```

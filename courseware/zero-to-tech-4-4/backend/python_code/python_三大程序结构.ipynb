1. 顺序结构 (Sequential Structure)
定义：程序按照代码从上到下、从左到右的顺序，一行一行地执行。
作用：这是程序最基础的执行逻辑，没有跳转，没有循环。
Java 对比：和 Java 完全一样，只是 Python 不需要在语句末尾加分号 ;

使用场景：几乎所有程序的初始化阶段、数据赋值、简单的线性计算。

```python
# 1. 准备数据
price = 100
quantity = 2

# 2. 计算总价
total = price * quantity

# 3. 打印结果
print(f"总价是: {total}元")
```

2. 分支结构 (Branching Structure)
定义：根据条件判断的真假，决定执行哪一段代码。
作用：让程序拥有“思考”和“选择”的能力。
Java 对比：Java 用 if...else if...else 和 switch，Python 用 if...elif...else（注意是 elif，不是 else if），且不需要小括号包裹条件。

用户权限校验（如：if user.role == 'admin':）
业务逻辑判断（如：库存不足时走不同的下单分支）
异常数据的拦截与处理

```python
score = 85

if score >= 90:
    print("优秀")          # 满足条件执行这里
elif score >= 60:
    print("及格")          # 不满足上面，但满足这里就执行
else:
    print("不及格")        # 都不满足执行这里
```

3. 循环结构 (Loop Structure)
定义：在满足特定条件时，重复执行某一段代码。
作用：让程序拥有“批量处理”的能力，告别复制粘贴。
Java 对比：Java 有 for(int i=0; i<n; i++) 和 while。Python 的循环更贴近自然语言，主要分为 for...in（增强型 for 循环）和 while。
① for...in 循环（最常用）

for：遍历数据库查询结果集、处理 API 返回的 JSON 数组、批量读取文件。
while：等待某个条件达成（如：轮询检查服务器状态、等待用户输入正确密码）。

```python
# for 循环
# 遍历列表（类似 Java 的 for(String name : names)）
names = ["Alice", "Bob", "Charlie"]
for name in names:
    print(f"你好, {name}")

# 循环指定次数（类似 Java 的 for(int i=0; i<5; i++)）
for i in range(5):
    print(f"当前数字是: {i}")
```

```python
#while循环
count = 0
while count < 3:
    print(f"循环第 {count} 次")
    count += 1  # Python 中没有 i++ 这种写法，必须写 count += 1
```

💡 新手避坑指南（Java 开发者专属提醒）
冒号不能忘：在 if、elif、else、for、while 的末尾，必须加英文冒号 :。
缩进即代码块：Java 用 {} 包裹代码块，Python 用缩进（通常是 4 个空格）。缩进不对，程序直接报错（IndentationError）。
没有 switch-case：Python 没有 switch 语句。如果有多个固定值的判断，可以用 if...elif...，或者使用字典映射（进阶技巧）。
没有 i++：Python 的自增必须写成 i += 1。

题目 1：简易 BMI 计算器（顺序 + 分支）
需求：
提示用户输入身高（米）和体重（千克）。
计算 BMI 值（公式：BMI = 体重 / 身高的平方）。
根据 BMI 值输出健康状态：
小于 18.5：偏瘦
18.5 ~ 24（不含24）：正常
24 ~ 28（不含28）：偏胖
28 及以上：肥胖
解题思路：
顺序：使用 input() 接收数据，注意 input() 返回的是字符串，需要用 float() 转换。
分支：使用 if...elif...else 结构，注意条件判断的顺序（从严格到宽松）。

```python
height = float (input("plese input your height:"));
weight = float (input("plese input your weight:"));
bmi = weight /( height ** 2);
# 分支结构：条件判断
if bmi < 18.5:
    status = "偏瘦"
elif bmi < 24:
    status = "正常"
elif bmi < 28:
    status = "偏胖"
else:
    status = "肥胖"

print(f"您的BMI值为: {bmi:.2f}，健康状态为: {status}")
```

题目 2：100以内的质数打印（循环 + 分支）
需求：
打印出 1 到 100 之间的所有质数（素数）。
(质数：大于1，且只能被1和它本身整除的自然数)
解题思路：
外层循环：遍历 2 到 100 的所有数字。
内层逻辑：判断当前数字是否为质数。假设它是质数，尝试用 2 到该数字的平方根之间的数去整除它。
分支：如果发现有能整除的数，说明不是质数，使用 break 跳出内层循环。
循环控制：如果内层循环正常结束（没有被 break 打断），说明是质数，打印出来。

```python

for num in range(2, 101):
    is_prime = True
    # 只需要除到平方根即可，提升效率
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break  # 找到因子，提前结束
            
    if is_prime:
        print(num, end=" ")
```

题目 3：猜数字小游戏（循环 + 分支综合）
需求：
程序随机生成一个 1 到 100 之间的整数。
让用户不断猜测，直到猜中为止。
每次猜测后，提示“猜大了”或“猜小了”。
猜中后，输出用户总共猜了多少次。
解题思路：
导入 random 模块生成随机数。
使用 while True 死循环，直到用户猜中时使用 break 退出。
使用变量 count 记录猜测次数。

```python
import random

num = random.randint(1,100)
count = 0 

print("=== 猜数字游戏开始 (1-100) ===")

while True:
    guess = int(input("请输入你猜的数字:"))
    count += 1
    if num == guess:
        print(f"猜对了,{guess}")
        print(f"一共猜了{count}次")
        break;
    elif num < guess:
        print("猜大了")
    elif num > guess:
        print("猜小了")   
        

# import random

# target = random.randint(1, 100)
# count = 0

# print("=== 猜数字游戏开始 (1-100) ===")
# while True:
#     guess = int(input("请输入你猜的数字："))
#     count += 1
    
#     if guess > target:
#         print("猜大了，再试试！")
#     elif guess < target:
#         print("猜小了，再试试！")
#     else:
#         print(f"恭喜你，猜对了！答案就是 {target}。")
#         print(f"你总共猜了 {count} 次。")
#         break  # 猜中后退出循环

```

题目 4：进阶挑战 - 打印等腰三角形（嵌套循环）
需求：
接收用户输入的行数 n，在控制台打印一个由 * 组成的等腰三角形。
例如输入 5，输出：
    *
   ***
  *****
 *******
*********

解题思路：
外层循环：控制行数（遍历 1 到 n）。
内层循环1：打印空格。第 i 行需要打印 n - i 个空格。
内层循环2：打印星号。第 i 行需要打印 2 * i - 1 个星号。
每行打印完毕后换行。


要解决这道题，我们需要像拆解机器一样，把整个三角形拆分成三个部分来思考：
1. 为什么要用嵌套循环？
因为打印图形本质上是一个二维平面的问题：
外层循环：控制行数（从上到下）。比如输入 5，外层循环就要执行 5 次。
内层循环：控制每一行的内容（从左到右）。在打印每一行时，我们需要先打印空格，再打印星号，这又需要循环。
2. 核心：寻找数学规律
假设用户输入的行数是 n = 5，我们逐行分析：

| 当前行 (i) | 需要打印的空格数 | 需要打印的星号(*)数 |
| :--- | :--- | :--- |
| 第 1 行 | 4 | 1 |
| 第 2 行 | 3 | 3 |
| 第 3 行 | 2 | 5 |
| 第 4 行 | 1 | 7 |
| 第 5 行 | 0 | 9 |






通过观察，我们可以得出两个极其重要的数学公式（假设当前是第 i 行，i 从 1 开始）：
空格数 = n - i （总行数减去当前行号）
星号数 = 2 * i - 1 （第1行1个，第2行3个... 这是一个奇数数列）
3. 解题思路（分三步走）
基于上面的规律，每一行的打印逻辑被严格规定为三步：
打印空格：循环 n - i 次，每次打印一个空格。
打印星号：循环 2 * i - 1 次，每次打印一个星号。
换行：这一行的空格和星号都打完后，必须执行一次 print() 换到下一行，否则所有字符都会挤在同一行。

```python
n = int(input("请输入三角形的行数："))

for i in range(1, n + 1):
    # 打印空格
    for _ in range(n - i):
        print(" ", end="")
    # 打印星号
    for _ in range(2 * i - 1):
        print("*", end="")
    # 换行
    print()
```

题目 1：简易 BMI 计算器（顺序 + 分支）
需求：
提示用户输入身高（米）和体重（千克）。
计算 BMI 值（公式：BMI = 体重 / 身高的平方）。
根据 BMI 值输出健康状态：
小于 18.5：偏瘦
18.5 ~ 24（不含24）：正常
24 ~ 28（不含28）：偏胖
28 及以上：肥胖
解题思路：
顺序：使用 input() 接收数据，注意 input() 返回的是字符串，需要用 float() 转换。
分支：使用 if...elif...else 结构，注意条件判断的顺序（从严格到宽松）。

```python
height = float(input("please input your height :"))
weight = float(input("please input your weight :"))

bmi = weight / (height ** 2);

if bmi < 18.5:
    print(f"your bmi is {bmi} , thinner")
elif bmi < 24:
    print(f"your bmi is {bmi} , normal")
elif bmi < 28:
    print(f"your bmi is {bmi} , litter fat")
elif bmi > 28:
    print(f"your bmi is {bmi} , fat")
```

题目 2：100以内的质数打印（循环 + 分支）
需求：
打印出 1 到 100 之间的所有质数（素数）。
(质数：大于1，且只能被1和它本身整除的自然数)
解题思路：
外层循环：遍历 2 到 100 的所有数字。
内层逻辑：判断当前数字是否为质数。假设它是质数，尝试用 2 到该数字的平方根之间的数去整除它。
分支：如果发现有能整除的数，说明不是质数，使用 break 跳出内层循环。
循环控制：如果内层循环正常结束（没有被 break 打断），说明是质数，打印出来。

> ⚠️ **注意事项：**
> - **`()` 圆括号**：代表元组，只遍历里面写死的几个元素。
> - **`range()` 函数**：代表数字区间，专门用来生成连续的整数序列。
> 
> 在写 `for` 循环遍历数字时，几乎 99% 的情况都需要配合 `range()` 使用。

```python
for num in range(2,100):
    is_prime = True;
    for i in range(2,int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break;
    
    if is_prime:
        print(num,end=" ")
```

题目 3：猜数字小游戏（循环 + 分支综合）
需求：
程序随机生成一个 1 到 100 之间的整数。
让用户不断猜测，直到猜中为止。
每次猜测后，提示“猜大了”或“猜小了”。
猜中后，输出用户总共猜了多少次。
解题思路：
导入 random 模块生成随机数。
使用 while True 死循环，直到用户猜中时使用 break 退出。
使用变量 count 记录猜测次数。

```python
import random
 
num = random.randint(1,100)

guess = int(input("please input guess number:"))

count = 0;
while True:
    guess = int(input("please input guess number:"))
          count += 1;
    if guess > num:
        print(f"bigger")
    elif guess < num:
        print(f"smaller")
    else:
        print(f"right count is :{count},num is {num}")
        break;


```

题目 4：进阶挑战 - 打印等腰三角形（嵌套循环）
需求：
接收用户输入的行数 n，在控制台打印一个由 * 组成的等腰三角形。
例如输入 5，输出：
    *
   ***
  *****
 *******
*********

```python
row = int(input("请输入行数："))

for i in range(1,row + 1 ):
    # 打印空格数量  row-i
    for j in range(1,row-i +1):
            print(" ", end="")  # 修改点：不换行，直接打印空格
    # 打印*数量 （1，3，5，7，9） 2*i -1
    for j in range(1,2*i -1 +1):
        print("*",end="")
    print()


# row = int(input("请输入行数："))

# for i in range(1, row + 1):
#     # 1. 打印空格数量：row - i
#     for j in range(row - i):
#         print(" ", end="")  # 修改点：不换行，直接打印空格
    
#     # 2. 打印 * 数量：2 * i - 1
#     for j in range(2 * i - 1):
#         print("*", end="")  # 修改点：不换行，直接打印星号
    
#     # 3. 一行结束后，必须换行
#     print() 
```

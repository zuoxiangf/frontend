
# Python 系统化学习大纲

## 一、Python 基础语法与数据类型（筑基）

- **环境搭建**：Python 解释器、虚拟环境（venv）、IDE 配置（PyCharm/VS Code）。
- **基本数据类型**：数值（int/float）、字符串（str）、布尔值（bool）。
- **核心数据结构**：列表（list）、元组（tuple）、字典（dict）、集合（set）及其底层原理与常用操作。
- **流程控制**：条件分支（if/elif/else）、循环结构（for/while）、跳转语句（break/continue/pass）。
- **Pythonic 语法糖**：列表推导式、字典推导式、切片操作（Slicing）、解包赋值（Unpacking）。

## 二、函数式编程与模块化（进阶）

- **函数基础**：函数定义与调用、位置参数、关键字参数、默认参数、可变参数（*args, **kwargs）。
- **变量作用域**：LEGB 规则、global 与 nonlocal 关键字。
- **函数的递归（新增）**：递归原理与经典案例（阶乘/斐波那契）、sys.getrecursionlimit() 深度限制、尾递归的局限性、如何用生成器/迭代替代深度递归。
- **高阶函数与匿名函数**：lambda 表达式、map()、filter()、reduce()、sorted() 自定义排序。
- **生成器与迭代器**：yield 关键字、生成器表达式、惰性求值机制（节省内存）。
- **模块与包管理**：import 机制、标准库使用（math/datetime/os/json）、pip 包管理工具、自定义模块与 `__init__.py`。

## 三、面向对象编程 OOP（核心）

- **类与对象基础**：类的定义、`__init__` 构造方法、实例属性与类属性、实例方法与类方法（@classmethod）。
- **三大特性**：封装、继承、多态（Python 的鸭子类型 Duck Typing）。
- **魔术方法（Dunder Methods）**：`__str__`、`__repr__`、`__eq__`、`__len__` 等，让自定义类支持内置操作。
- **高级 OOP 特性**：属性装饰器（@property）、抽象基类（ABC）、多重继承与 MRO（方法解析顺序）。

## 四、高级特性与工程化（高阶）

- **异常处理**：try/except/else/finally 结构、自定义异常类、上下文管理器（with 语句）。
- **文件与 IO 操作**：文本文件读写、CSV/JSON 数据处理、目录与路径管理（pathlib）。
- **正则表达式**：re 模块核心语法、数据提取与文本清洗。
- **装饰器（Decorator）**：闭包原理、函数/类装饰器、带参数的装饰器、functools.wraps。
- **并发编程基础**：多线程（threading）、多进程（multiprocessing）、协程（asyncio）初步概念。

## 五、实战应用与方向拓展（落地）

- **Web 开发方向**：Flask / Django 框架、RESTful API 设计、数据库 ORM（SQLAlchemy）。
- **数据分析方向**：NumPy（数值计算）、Pandas（数据处理）、Matplotlib/Seaborn（数据可视化）。
- **自动化与爬虫**：Requests + BeautifulSoup、Selenium 网页自动化、定时任务调度。

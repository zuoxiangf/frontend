
# Python 类的定义与实例化

## 一、 类的定义与实例化

### 1. 定义类（Class Definition）
- **核心语法**：使用 `class` 关键字定义类，类名通常遵循大驼峰命名法（PascalCase）。
- **Java 对比**：Python 的类不需要写在特定的文件里（一个 `.py` 文件可以包含多个类），也不需要访问修饰符（如 `public class`）。

### 2. 构造方法 `__init__`
- **是什么**：当实例化一个类时，Python 会自动调用 `__init__` 方法来初始化对象的属性。
- **Java 对比**：相当于 Java 中的构造方法（与类同名的方法）。
- **注意**：它是双下划线开头和结尾的“魔术方法（Dunder Method）”。

### 3. 实例化（Instantiation）
- **核心语法**：直接调用类名，就像调用一个函数一样（例如 `dog = Dog()`）。
- **Java 对比**：Java 必须使用 `new` 关键字（如 `Dog dog = new Dog()`）。Python 认为 `new` 是多余的，直接调用类名即可。

### 4. 核心关键字 `self`（重点！）
- **是什么**：`self` 代表**当前实例本身**。在类的方法中，必须把 `self` 作为**第一个参数**。
- **Java 对比**：相当于 Java 中的 `this`。
- **颠覆认知的点**：
    - 在 Java 中，`this` 是隐式的，你不需要在方法参数里写 `this`。
    - 在 Python 中，`self` 在**定义方法时必须显式写出**，但在**调用方法时是隐式传入的**。

---

## 二、 完整代码示例

我们用一个“银行账户”的例子来直观感受：

```python
# 1. 定义类
class BankAccount:
    # 2. 构造方法：初始化账户名和余额
    # 注意：self 必须是第一个参数
    def __init__(self, owner, balance=0):
        self.owner = owner      # 将传入的参数绑定到当前实例的属性上
        self.balance = balance

    # 3. 普通方法：存款
    # 同样，self 必须是第一个参数
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"{self.owner} 成功存入 {amount} 元，当前余额: {self.balance}")

    # 4. 普通方法：获取信息
    def get_info(self):
        return f"账户: {self.owner}, 余额: {self.balance}"

# 5. 实例化（不需要 new 关键字！）
account1 = BankAccount("张三", 1000)
account2 = BankAccount("李四")  # balance 使用默认值 0

# 6. 调用方法（不需要传 self，Python 会自动把 account1 传给 self）
account1.deposit(500)  
print(account2.get_info()) 
```

---

## 三、 底层执行过程解析（帮你彻底搞懂 self）

当你写下 `account1.deposit(500)` 时，Python 底层到底做了什么？

1. Python 发现 `account1` 是一个 `BankAccount` 的实例。
2. 它去 `BankAccount` 类中找到 `deposit` 方法。
3. **关键一步**：Python 自动把 `account1` 本身作为第一个参数传入，相当于执行了 `BankAccount.deposit(account1, 500)`。
4. 在 `deposit` 方法内部，`self` 就接收了这个 `account1`，所以 `self.balance += amount` 实际上修改的是 `account1` 的余额。

## 💡 Java 开发者避坑指南

1. **千万别漏掉 `self`**：这是新手最容易犯的错。定义方法时如果不写 `self`，调用时会报错：`TypeError: deposit() takes 1 positional argument but 2 were given`。
2. **属性的动态绑定**：Java 中类的属性必须在类顶部提前声明。但在 Python 中，你**不需要提前声明属性**，只要在 `__init__` 或其他方法中通过 `self.xxx = yyy` 直接赋值即可。甚至你可以在实例化之后，动态给对象添加新属性（虽然不推荐这么做）。
3. **不要混淆类属性和实例属性**：
    - 写在类里面、方法外面的变量（如 `class Dog: species = '犬科'`）是**类属性**，所有实例共享（相当于 Java 的 `static` 变量）。
    - 写在 `__init__` 里带 `self.` 的变量是**实例属性**，每个对象独有。

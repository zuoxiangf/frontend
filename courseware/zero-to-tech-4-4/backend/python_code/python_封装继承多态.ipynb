
# Python 的封装、继承与多态（鸭子类型 Duck Typing）

作为 Java 开发者，你已经习惯了强类型、接口和严格的访问修饰符。但在 Python 中，这三大特性的实现方式非常“Pythonic”，充满了灵活性和动态性。

## 一、 封装（Encapsulation）：约定俗成的隐私

**Java 认知**：使用 `private`、`protected`、`public` 关键字严格控制访问权限。
**Python 认知**：Python 崇尚“我们都是成年人”，没有强制的访问控制，而是通过**命名约定**来实现封装。

### 1. 访问控制的三种级别

- **公有成员（Public）**：默认情况，任何地方都可以访问（如 `self.name`）。
- **受保护成员（Protected）**：以**单下划线**开头（如 `self._age`）。这只是一个“君子协定”，提示外部开发者“这是内部使用的，请别乱动”，但技术上依然可以直接访问。
- **私有成员（Private）**：以**双下划线**开头（如 `self.__salary`）。这会触发 Python 的**名称修饰（Name Mangling）**机制，解释器会把它重命名为 `_ClassName__salary`，从而防止外部直接访问。

### 2. 高级封装利器：`@property`

在 Java 中，为了封装，你会写一堆 `getAge()` 和 `setAge()`。在 Python 中，使用 `@property` 装饰器，你可以**像访问属性一样调用方法**，同时还能在内部做数据校验。

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self._age = age  # 约定为受保护属性

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0 or value > 150:
            raise ValueError("年龄不合法！")
        self._age = value

p = Person("Alice", 25)
p.age = 30  # 像赋值一样触发 setter，自动进行校验
```

## 二、 继承（Inheritance）：代码复用与 MRO

**Java 认知**：单继承（`extends`），多实现（`implements`）。
**Python 认知**：支持**多继承**，且没有 `new` 关键字。

### 1. 基础继承与 `super()`

子类可以继承父类的所有公有属性和方法，并使用 `super()` 调用父类的初始化或方法。

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        pass

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # 调用父类构造方法
        self.breed = breed

    def speak(self):
        return f"{self.name} says Woof!"
```

### 2. 多继承与 MRO（方法解析顺序）

Python 允许一个子类继承多个父类（如 `class C(A, B):`）。当多个父类有同名方法时，Python 通过 **MRO（Method Resolution Order）** 机制，按照**从左到右、广度优先（C3线性化算法）**的顺序查找。

- 你可以通过 `print(C.__mro__)` 查看具体的查找顺序。
- **避坑提示**：实际工程中尽量**慎用多继承**，推荐使用 **Mixin（混入）模式**或**组合**来替代，避免“菱形继承”带来的复杂性。

## 三、 多态（Polymorphism）：鸭子类型（Duck Typing）

这是 Python 与 Java **差异最大、也最优雅**的地方！

**Java 认知**：多态依赖于严格的继承体系或接口实现。你必须声明 `Animal a = new Dog()`，或者函数参数必须明确为 `Animal` 类型。
**Python 认知**：**鸭子类型（Duck Typing）**——“如果它走起路来像鸭子，叫起来像鸭子，那么它就是鸭子”。

### 1. 什么是鸭子类型？

在 Python 中，**不关心对象的具体类型或继承关系，只关心它是否具备所需的行为（方法）**。只要对象实现了特定的方法，就可以被传入函数中使用。

```python
# 没有任何继承关系，甚至不是动物！
class Airplane:
    def speak(self):
        return "I am an airplane!"

# 多态函数：不限制参数类型，只要求有 speak() 方法
def make_noise(thing):
    print(thing.speak())

dog = Dog("Buddy", "Husky")
plane = Airplane()

make_noise(dog)    # 输出: Buddy says Woof!
make_noise(plane)  # 输出: I am an airplane! (完全合法！)
```

### 2. 鸭子类型的优缺点

- **优点**：极其灵活，解耦彻底。你可以随时替换组件，只要接口（方法名）一致即可，非常适合数据处理和插件化架构。
- **缺点**：失去了编译期的类型安全检查。如果传入了一个没有 `speak()` 方法的对象，只有在**运行时**才会抛出 `AttributeError`。

### 3. Python 的软约束：抽象基类（ABC）

如果你既想要鸭子类型的灵活，又想要 Java 接口的强制约束，可以使用 `abc` 模块定义抽象基类，强制子类必须实现某些方法。

## 💡 核心总结（Java vs Python OOP）

| 特性 | Java 的做法 | Python 的做法 |
| :--- | :--- | :--- |
| **封装** | `private/public` 强控制 | `_` 约定，`__` 名称修饰，`@property` 优雅拦截 |
| **继承** | 单继承，`super()` | 多继承，MRO 查找顺序，`super()` |
| **多态** | 强类型，依赖接口/继承体系 | **鸭子类型**，关注行为而非类型，运行时动态绑定 |

```python
#1、需求1：根据输入的直角边的边长，打印等腰直角三角形
a = int (input("输入的直角边的边长:"))
print(f"输入的边长是：{a}")

for i in range(1,a+1): #行
    for j in range(1,i+1): #列
        print(f"{j}" ,end= "\t")
    print()

```

```python
#3、需求3：打印国际象棋棋盘
def printStr(i,j):
    for k in range(1,j+1):
        if  i % 2 == 1 :
            if k % 2 == 1 :
                print(f"■" ,end= " ")
            else :
                print(f"□", end= " ")    
        else :        
            if k % 2 != 1 :
                print(f"■" ,end= " ")
            else :
                print(f"□", end= " ")   
        
for i in range(1,9): #行
    printStr(i,9)
    print()

```

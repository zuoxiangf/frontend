
# Python Lambda 表达式与函数式编程核心特性

## 1. Lambda 表达式：匿名函数

**是什么**：一种极简的、没有名字的函数写法。
**Java 对比**：Java 的 `(a, b) -> a + b`。
**语法**：`lambda 参数列表: 返回值表达式`（只能写一行，隐式返回，不需要 `return`）。

**代码示例**：
```python
# 传统函数
def add(x, y):
    return x + y

# Lambda 表达式（一行搞定）
add_lambda = lambda x, y: x + y

print(add_lambda(3, 5))  # 输出：8
```

**使用场景**：作为高阶函数（如 `map`, `filter`）的参数时，不需要单独去定义一个函数。

---

## 2. map()：批量转换

**是什么**：将一个函数应用到可迭代对象的**每一个元素**上，返回一个新的迭代器。
**Java 对比**：`stream().map()`。

**代码示例**：
```python
nums = [1, 2, 3, 4]

# 需求：把所有数字平方
# map + lambda 写法：
squared = list(map(lambda x: x ** 2, nums))

print(squared)  # 输出：[1, 4, 9, 16]
```

**⚠️ 注意**：`map()` 返回的是惰性迭代器，必须用 `list()` 包一下才能看到结果。

---

## 3. filter()：条件过滤

**是什么**：将一个返回布尔值的函数应用到可迭代对象上，保留返回 `True` 的元素。
**Java 对比**：`stream().filter()`。

**代码示例**：
```python
nums = [1, 2, 3, 4, 5, 6]

# 需求：提取所有偶数
evens = list(filter(lambda x: x % 2 == 0, nums))

print(evens)  # 输出：[2, 4, 6]
```

---

## 4. reduce()：累积计算

**是什么**：将一个双参数函数依次作用于序列的元素，将结果不断累积，最终得出一个单一的值。
**Java 对比**：`stream().reduce()`。

**代码示例**：
```python
from functools import reduce  # 注意：reduce 不在内置函数中，需要导入

nums = [1, 2, 3, 4]

# 需求：计算所有数字的乘积 (1 * 2 * 3 * 4)
# x 是累积结果，y 是当前元素
product = reduce(lambda x, y: x * y, nums)

print(product)  # 输出：24
```

**执行过程**：`(1*2)=2` -> `(2*3)=6` -> `(6*4)=24`。

---

## 5. sorted()：自定义排序

**是什么**：对可迭代对象进行排序，返回一个**新列表**（不修改原列表）。
**Java 对比**：`Collections.sort(list, comparator)`。
**核心参数**：`key` 参数接收一个函数，`sorted()` 会根据这个函数的返回值来决定排序规则。

**代码示例**：
```python
# 场景1：按绝对值排序
nums = [-3, 1, -4, 2]
sorted_nums = sorted(nums, key=lambda x: abs(x))
print(sorted_nums)  # 输出：[1, 2, -3, -4]

# 场景2：对字典列表排序（极其常用！）
students = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 20},
    {"name": "Charlie", "age": 30}
]

# 按年龄从小到大排序
by_age = sorted(students, key=lambda s: s["age"])

# 按年龄从大到小排序（加 reverse=True）
by_age_desc = sorted(students, key=lambda s: s["age"], reverse=True)
```

## 💡 核心总结与 Java 对比

| Python 写法 | Java Stream 等价写法 | 作用 |
| :--- | :--- | :--- |
| `map(lambda x: x*2, lst)` | `stream().map(x -> x*2)` | 转换数据 |
| `filter(lambda x: x>0, lst)` | `stream().filter(x -> x>0)` | 过滤数据 |
| `reduce(lambda x,y: x+y, lst)` | `stream().reduce((x,y) -> x+y)` | 聚合计算 |
| `sorted(lst, key=lambda x: x.age)` | `stream().sorted(Comparator.comparing(User::getAge))` | 自定义排序 |

**Pythonic 进阶提示**：
虽然 `map` 和 `filter` 很强大，但在 Python 中，**列表推导式（List Comprehension）** 往往比它们更受欢迎，因为可读性更好！
例如：`list(map(lambda x: x**2, nums))` 完全可以写成 `[x**2 for x in nums]`。
